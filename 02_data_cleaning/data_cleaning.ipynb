# 1. Data Types conversion of DataFrames

```python
import pandas as pd
import numpy as np
```

#### Loading raw data file for Pandas to read it

```python
run_time = pd.read_excel('data/Run Times.xlsx')
run_time
```

#### Checking the DataTypes of each column

```python
run_time.dtypes
```

### Objectives:

- The `Warm Up Time` column is currently of dtype `object` because the second row contains mixed data (3 min). We need to clean the values and convert the column to a float.
- The `Fee` column is stored as a `string` because it includes the `$` symbol. We need to remove the `$` symbol and then convert the column to a float.



```python
run_time
```

```python
# Working on Fee Column as per our objective

pd.to_numeric(run_time.Fee.str.replace('$',''))
```

```python
# Working on `Warm Up Time` Column as per our objective

pd.to_numeric(run_time['Warm Up Time'], errors='coerce')
```

```python
# now column's data type converted to float64 but 2nd column value changed to NaN
# So first we need to remove the `min` from the Warm Up Time (google: str.replace nan).DS_Store

run_time['Warm Up Time'].str.replace(' min', '')
```

```python
# above method also fails as now the problematic column corrected
# but other column value changed to Nan and still data type is object.

# Therefor we can use astype() to replace the string min.

pd.to_numeric(run_time['Warm Up Time'].astype('str').str.replace(' min', ''))
```

```python
# change the Rain column from boolean to int

run_time.Rain.astype('int')
```
